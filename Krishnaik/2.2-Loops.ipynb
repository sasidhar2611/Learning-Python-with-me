```python
## While loop

## The while loop continues to execute as long as the condition is True

count = 0
while count<5:
    print(count)
    count += 1
```

```python
count = 0

while count % 2 == 0:
    print(count)
    count += 1
```

```python
## Loop control Statements

## Break
## The break statment exists the loop prematurely

## break Statement

for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
## continue
## The continue statment skips the current iteration and continues with the next

for i in range(10):
    if i % 2 == 0:
        continue
    print(i)

## skips the even numbers and print the odd ones
```

```python
## pass
## The pass statement is a null operation, it does nothing

for i in range(5):
    if i == 3:
        print("The number here is ", i)
        pass
    print(i)
```

```python
## Nested For and while Loops
## A loop inside a loop
## f statement in the print is a formating option, it is used for multilpe variable printing easily

for i in range(3):
    for j in range(2):
        print(f"i : {i}, j : {j}")
```

```python
## Examples - Calculate the sum of frist N natural numbers using a while and for loop

## while loop

n=int(input("Enter the integer: "))
sum = 0
count = 1

while count <= n:
    sum += count
    count += 1

print("Sum of first N natural numbers will be equal to ",sum)
```

```python
## for loop

n=int(input("Enter the integer: "))
sum = 0

for i in range(11):
    sum += i

print(sum)
```

```python
## Example - Prime numbers between 1 and hundered

for num in range(1,101):
    if num>1:
        for i in range(2, num):
            if num%i == 0:
                break
        else:
            print(num)
```

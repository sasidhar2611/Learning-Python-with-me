```python
## Why functions?
num = 24
if num%2 == 0:
    print("The number is even")
else:
    print("The number is odd")
```

```python
def even_or_odd(num):
    '''This function find even or odd'''
    if num % 2 == 0:
        print("The number is even")
    else:
        print("The number is odd")
```

```python
## call this fucntion

n = int(input("Enter your Number: "))
even_or_odd(n)
```

```python
## function with multiple parameters

def add(a,b):
    '''This function add 2 number'''
    return a+b

x = int(input("Enter the first number"))
y = int(input("Enter the Second number"))

result = add(x,y)
print(result)
```

```python
## Default parameters

def greet(name):
    print(f"Hello {name}! Well come to the paradise")

greet("Sasidhar")
```

```python
def greet(name="Guest"):    ## default parameter here is the guest
    print(f"Hello {name}! Well come to the paradise")

greet()
```

```python
## Variable length Arguments
## Positional and Keywords Argument

def print_numbers(*args): # *args can be anything like *sasidhar
    for number in args:
        print(number)
```

```python
print_numbers(1,2,3,4,5,6,7,8,"sasidhar")
```

```python
## Keywords argument

def print_detail(**kwargs):
    for key,value in kwargs.items():
        print(f"{key} : {value}")
```

```python
print_detail(name="Sasidhar", age = 32, country ="India")
```

```python
def print_detail(*args,**kwargs):
    for val in args:
        print(f"Positional Argument : {val}")
    
    for key,value in kwargs.items():
        print(f"{key} : {value}")
```

```python
print_detail(1,2,3,4,"Sasi", name="Sasidhar", age = 20, country ="India")
```

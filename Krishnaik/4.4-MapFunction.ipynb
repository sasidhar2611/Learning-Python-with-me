```python
## The map() function in Python

# The map() function applies a given function to all items in an input list (or any other iterable) and retruns a map object(an iterator).
# This is particularly useful fro transforming data in a list comprehensively.
```

```python
def square(x):
    return x*x

square(10)
```

```python
numbers=[1,2,3,4,5,6,7,8]

list(map(square, numbers))                  ## two input parameters are used, 1st-function name, 2nd-iterables, applies the input function to all the iterables in the list
```

```python
## Lambda function with map
numbers=[1,2,3,4,5,6,7,8]

list(map( lambda x:x*x, numbers))
```

```python
## Map multiple iterables

numbers1=[1,2,3]
numbers2=[4,5,6]

added_numbers=list(map(lambda x,y:x+y, numbers1, numbers2))
print(added_numbers)
```

```python
#map() to convert list of string to integers

str_numbers = ['1', '2', '3', '4', '5']
int_numbers = list(map(int, str_numbers))     ## int is a function to type cast it

print(int_numbers)
```

```python
word=['apple', 'banana', 'cherry']
upper_word = list(map(str.upper, word))
print(upper_word)
```

```python
def get_name(person):
    return person['name']

people = [
    {'name':'Sasidhar', 'age':19},
    {'name':'Harsha', 'age':20}
]
list(map(get_name, people))
```

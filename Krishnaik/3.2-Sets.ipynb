```python
## Sets

## Sets re built-in data type in python used to store collections of unique
#  items. They are unordered, meaning that the elements do not follow a
#  specific order, and they do not allow duplicate elements. Sets are useful
#  for membership test, eliminating duplicate enteries, and performing
#  mathematical set operations like union, intersection, difference and 
# symmetric differnece
```

```python
## create a set

my_set = {1,2,3,4,5}
print(my_set)
print(type(my_set))
```

```python
my_empty_set = set()
print(type(my_empty_set))
```

```python
my_empty_set = set([1,2,3,4,5,6])
print(type(my_empty_set))
```

```python
my_set = set([1,2,3,6,5,4,4,5,6])
print(my_set)
```

```python
## Basic Set operations
## adding and removing elements

my_set.add(7)
print(my_set)
my_set.add(7)
print(my_set)
```

```python
## remove the elements from a set

my_set.remove(3)
print(my_set)
```

```python
# my_set.remove(10)
# print(my_set)
```

```python
my_set.discard(11)
print(my_set)
```

```python
## pop method

removed_element = my_set.pop()
print(removed_element)
print(my_set)
```

```python
## clear all the elements
my_set.clear()
print(my_set)
```

```python
## Set membership test

my_set = {1,2,3,4,5}
print(3 in my_set)
print(10 in my_set)
```

```python
## Mathematical operation

set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}


## Union

union_set = set1.union(set2)
print(union_set)


## Intersection

intersection_set = set1.intersection(set2)
print(intersection_set)

set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1,2,3,4,5,6}
set2 = {4,5,6,7,8,9}

## Difference
print(set1.difference(set2))
```

```python
set2.difference(set1)
```

```python
## Symmetric Difference - intersection elements gets removed

set1.symmetric_difference(set2)
```

```python
## Sets Methods

set1 = {1,2,3,4,5}
set2 = {3,4,5}

## is subset

print(set1.issubset(set2))  ## does set 2 have all elements of set 1

print(set1.issuperset(set2)) ## does set 1 have all the elements of set 2
```

```python
lst = [1,2,2,3,4,4,5]  ## delete duplicates from list using set **A small shortcut

set(lst)
```

```python
### counting unique word in text

text ="In ths tutorial we are discussing about sets"
words = text.split()

## convert list of words to set to get unique words

unique_words = set(words)
print(unique_words)
print(len(unique_words))
```

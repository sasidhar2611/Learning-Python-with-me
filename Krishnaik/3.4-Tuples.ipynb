```python
## Packing and unpacking tuples

packed_tuple = 1,"hello", 3.14 ## when ever items are listed like this with a coma seperation, they are automatically packed in tuple
print(packed_tuple)
```

```python
a,b,c = packed_tuple ##unpacking a tuple

print(a)
print(b)
print(c)
```

```python
## Unpacking with *

numbers=(1,2,3,4,5,6)
first, *middle, last = numbers
print(first)
print(middle)  ## all ements from where it stated to the last element will be picked up and put into this variable
print(last)
```

```python
## Nested tuples
## Nested List
lst=[[1,2,3,4],[6,7,8,9],[1,"Hello",3,14,"c"]]
lst[0][2]  ## increasing number of indexes to go deep for that element
lst[0][0:3]
```

```python
lst=[[1,2,3,4],[6,7,8,9],(1,"Hello",3,14,"c")]
lst[2][0:3]
```

```python
nested_tuple = ((1,2,3),("a","b","c"),(True, False))

## acess elements inside a tuple

print(nested_tuple[0])
print(nested_tuple[1][2])
```

```python
## iterating over nested tuples

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item, end=" ")
    print()
```
